```python
import xarray as xr
import matplotlib.pyplot as plt
```

### Evolution of temperature profile with and without OTEC (and some corner cases)

```python
plt.figure(figsize=(20,5))

exps = ["without_otec", "with_otec_shallow", "with_otec", "with_otec_no-regrid"]
long_names = {
    "without_otec": "Reference (OTEC turned off)",
    "with_otec_shallow": "OTEC turns off if shallower than intake",
    "with_otec_no-regrid": "OTEC turned on (but ALE remap off)",
    "with_otec": "OTEC turned on ($w_{CW} = 10^{-4}$ m/s)"
}

for i, exp in enumerate(exps, start=1):
    plt.subplot(1,len(exps),i)
    ds = xr.open_dataset(f"../{exp}/prog_z.nc", decode_times=False)
    for t in ds['Time'][4::5]:
        ds['temp'].sel(Time=t).mean(['xh', 'yh']).plot(label=t.to_masked_array(), y='z_l')
    plt.grid(True)

    plt.ylim(1200, 0)
    plt.xlim(0, 30)
    plt.title(f"{long_names[exp]}")
    plt.ylabel("Depth [m]")
    plt.xlabel(r"Temperature [$\degree$C]")
    plt.legend();
```
